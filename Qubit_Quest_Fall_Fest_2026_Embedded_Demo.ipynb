## 4 · Independent Qiskit experiments

The following demonstration cells reproduce the concepts in the repository’s `notebooks/judge_demo.ipynb`: interference, entanglement, search, QFT, optimization, and toy variational energy. They provide a runnable technical reference alongside the web app.

**Setup:** Install the Python dependencies specified in [`services/api/requirements.lock.txt`](https://github.com/minhajuddinm/qubit-quest/blob/main/services/api/requirements.lock.txt). These are **local ideal simulations**, not runs on quantum hardware. If you only want to view the website, skip this section.

### 4.1 · Load Qiskit and numerical tools

```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector, partial_trace, SparsePauliOp
from qiskit.circuit.library import QFTGate
from qiskit.primitives import StatevectorSampler
import numpy as np
```

### 4.2 · Superposition and interference

Predict `H|0⟩`, then `H²|0⟩`. Equal measurement probabilities do not imply a classical mixture.

```python
qc = QuantumCircuit(1)
qc.h(0)
print('After H:', Statevector(qc).probabilities_dict())
qc.h(0)
print('After HH:', Statevector(qc).probabilities_dict())
assert np.isclose(Statevector(qc).probabilities()[0], 1)
```

### 4.3 · Bell correlations and reduced states

Bitstrings are `q1q0`. Tracing out qubit 1 leaves a maximally mixed qubit 0.

```python
bell = QuantumCircuit(2)
bell.h(0)
bell.cx(0, 1)
print(Statevector(bell).probabilities_dict())
print(partial_trace(Statevector(bell), [1]))
measured = bell.copy()
measured.measure_all()
print(StatevectorSampler(seed=7).run([measured], shots=1024).result()[0].data.meas.get_counts())
```

### 4.4 · Grover search: mark `11` and amplify

For four candidates and one marked item, one iteration is sufficient. The oracle is supplied; it is not a free database lookup.

```python
grover = QuantumCircuit(2)
grover.h([0, 1])
grover.cz(0, 1)
grover.h([0, 1])
grover.x([0, 1])
grover.cz(0, 1)
grover.x([0, 1])
grover.h([0, 1])
print(Statevector(grover).probabilities_dict())
assert np.isclose(Statevector(grover).probabilities()[3], 1)
```

### 4.5 · Quantum Fourier transform

Inspect complex amplitudes, not just measurement histograms.

```python
qft = QuantumCircuit(2)
qft.x(0)
qft.append(QFTGate(2), [0, 1])
print(Statevector(qft).data)
assert np.allclose(Statevector(qft).data, [0.5, 0.5j, -0.5, -0.5j])
```

### 4.6 · Optimization: one-edge MaxCut

`01` and `10` cut the edge. Sweep `gamma` while fixing `beta = -π/8`.

```python
best = (0, 0)
for gamma in np.linspace(0, 2*np.pi, 101):
    circuit = QuantumCircuit(2)
    circuit.h([0, 1])
    circuit.rzz(gamma, 0, 1)
    circuit.rx(-np.pi/4, [0, 1])
    p = Statevector(circuit).probabilities()
    score = p[1] + p[2]
    if score > best[1]:
        best = (gamma, score)
print('Best angle and expected cut:', best)
assert np.isclose(best[1], 1)
```

### 4.7 · Variational energy: chemistry precursor

This is a **toy Hamiltonian**, `−ZI − IZ + 0.5XX`, in arbitrary units. It is **not** an ab initio molecule. Real chemistry requires molecular geometry, a basis, an active space, a mapping, and validated reference energies.

```python
hamiltonian = SparsePauliOp.from_list([('ZI', -1), ('IZ', -1), ('XX', .5)])
exact = np.linalg.eigvalsh(hamiltonian.to_matrix())[0]
energies = []
for theta in np.linspace(0, 2*np.pi, 201):
    trial = QuantumCircuit(2)
    trial.ry(theta, 0)
    trial.cx(0, 1)
    energies.append(Statevector(trial).expectation_value(hamiltonian).real)
print('Trial minimum:', min(energies), 'Exact ground:', exact)
assert min(energies) >= exact - 1e-10
```
